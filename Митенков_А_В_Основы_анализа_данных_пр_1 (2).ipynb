```python

class Chocolate():
  def __init__(self,cost:float, addition :str | None=None): # использую аннотацию типов из typing, вместо None можно указать Optional
    if cost>0: # проверяю по второму условию
      self.cost=float(cost) # привожу к типу
      self.addition =addition # не привожу к типу, т.к когда приводил были проблемы проверки в  show_my_chocolate
    else:
      raise ValueError("Цена не может быть меньше нуля")
  def show_my_chocolate(self):
    if self.addition is not None:
      return print(f"Шоколадный батончик, добавка: {self.addition}")
    else:
      return print("Обычный шоколадный батончик")
  def __str__(self):
     return f'Шоколадный батончик, {self.cost} рублей' # цена с будет возвращаться с плавающей точкой, конкретного требования на отображения не было

#Примеры
banana=Chocolate(500,'banana')
print(banana)
banana.show_my_chocolate()
print(banana.cost)

""" Вызов ошибки
honey=Chocolate(-1,'honey')
print(honey)
honey.show_my_chocolate()
print(honey.cost)
"""

nuts=Chocolate(100)
print(nuts)
nuts.show_my_chocolate()
print(nuts.cost)
```
